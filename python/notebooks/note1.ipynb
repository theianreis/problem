```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import PchipInterpolator

# ============================================================
# PADRÃO DOS GRÁFICOS (igual para todos os experimentos)
# ============================================================
Y_MIN, Y_MAX, Y_STEP = -17, 32, 2
TEMP_MAX_GELADEIRA = 5      # °C
TEMP_MAX_FREEZER = -15      # °C


def para_minutos(h):
    hh, mm = h.split(":")[:2]
    return int(hh) * 60 + int(mm)


def tempo_decorrido(horarios):
    """Minutos desde o primeiro horário, tratando a virada da meia-noite."""
    mins, dia, ant = [], 0, None
    for h in horarios:
        m = para_minutos(h)
        if ant is not None and m < ant:
            dia += 1440
        mins.append(m + dia)
        ant = m
    mins = np.array(mins)
    return mins - mins[0]


def plotar_experimento(t, temp, titulo, assintota, rotulo_assintota=None,
                       xlabel="Tempo (min)", arquivo=None):
    t = np.array(t, dtype=float)
    temp = np.array(temp, dtype=float)

    fig, ax = plt.subplots(figsize=(11, 6.5))

    # Curva suave passando pelos pontos
    if len(t) >= 3:
        t_suave = np.linspace(t.min(), t.max(), 400)
        ax.plot(t_suave, PchipInterpolator(t, temp)(t_suave),
                color="#1f77b4", linewidth=2, zorder=2, label="Curva")

    # Pontos medidos
    ax.scatter(t, temp, color="#1f77b4", s=45, zorder=3, label="Medições")

    # Destaque: temperatura inicial e final
    ax.scatter(t[0], temp[0], color="green", s=160, zorder=4,
               edgecolors="black", label=f"Inicial: {temp[0]:.1f} °C")
    ax.scatter(t[-1], temp[-1], color="red", s=160, zorder=4,
               edgecolors="black", label=f"Final: {temp[-1]:.1f} °C")
    ax.annotate(f"{temp[0]:.1f} °C", (t[0], temp[0]),
                textcoords="offset points", xytext=(10, 10),
                fontweight="bold", color="green")
    ax.annotate(f"{temp[-1]:.1f} °C", (t[-1], temp[-1]),
                textcoords="offset points", xytext=(-10, 12),
                fontweight="bold", color="red", ha="right")

    # Assíntota (linha pontilhada)
    ax.axhline(assintota, color="gray", linestyle=":", linewidth=2,
               label=rotulo_assintota or f"Estabilização: {assintota} °C")

    # Eixo Y padronizado: -17 a 32, passos de 2
    ax.set_ylim(Y_MIN, Y_MAX)
    ax.set_yticks(np.arange(Y_MIN, Y_MAX + 1, Y_STEP))

    ax.set_xlabel(xlabel)
    ax.set_ylabel("Temperatura (°C)")
    ax.set_title(titulo, fontweight="bold")
    ax.grid(True, alpha=0.3)
    ax.legend(loc="center right")

    plt.tight_layout()
    if arquivo:
        plt.savefig(arquivo, dpi=200)
    plt.show()


# ============================================================
# EXPERIMENTO 3: CHOCOLATE - DESCONGELAMENTO NA GELADEIRA
# ============================================================
horarios = ["07:00", "07:20", "07:40", "08:00", "08:20", "08:40", "09:00",
            "09:20", "09:40", "10:00", "10:20", "10:40", "11:00", "11:20",
            "11:40", "12:00", "12:20", "12:40"]

t_chocolate_desc_geladeira = tempo_decorrido(horarios)   # 0, 20, ..., 340

temp_chocolate_desc_geladeira = np.array([
    -15.5, -9.3, -5, -2, 0.2, 1.6, 2.7, 3.4, 3.9, 3.9,
    4.1, 4.2, 4.2, 4.4, 4.6, 5.1, 5, 5
])

plotar_experimento(
    t_chocolate_desc_geladeira,
    temp_chocolate_desc_geladeira,
    titulo="Experimento do Chocolate – Descongelamento na Geladeira",
    assintota=5,
    rotulo_assintota="Estabilização (medida): 5,0 °C",
    xlabel="Tempo (min)",
    arquivo="chocolate_descongelamento_geladeira.png",
)
```
