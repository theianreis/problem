
# Arquitetura de Modelagem Computacional

## 1. Visão geral

**Objetivo:** transformar um fenômeno físico em um modelo matemático,
desenvolver uma aproximação numérica e avaliar a confiabilidade dos resultados.

```mermaid
flowchart TD
    A["1. Definição do problema"] --> B["2. Modelagem física"]
    B --> C["3. Formulação matemática"]
    C --> D["4. Análise do modelo"]
    D --> E["5. Seleção do método numérico"]
    E --> F["6. Discretização"]
    F --> G["7. Implementação computacional"]
    G --> H["8. Simulação"]
    H --> I["9. Verificação numérica"]
    I --> J["10. Validação física"]
    J --> K["11. Análise e documentação"]

    I -->|Erro numérico elevado| E
    J -->|Modelo inadequado| B
    K -->|Novo estudo| A
```

---

## 2. Definição do problema

### Objetivo
Determinar quais fenômenos serão estudados e quais grandezas
precisam ser calculadas.

### Atividades
- Definir o sistema.
- Identificar entradas e saídas.
- Definir o domínio espacial e/ou temporal.
- Estabelecer objetivos da simulação.
- Definir critérios de aceitação.

### Entregáveis
- Descrição do problema.
- Grandezas de interesse.
- Domínio de análise.
- Critérios de avaliação.

---

## 3. Modelagem física

### Objetivo
Representar o fenômeno com base em princípios físicos.

### Atividades
- Identificar componentes e suas interações.
- Estabelecer hipóteses simplificadoras.
- Identificar leis de conservação.
- Definir propriedades físicas.
- Determinar condições iniciais e de contorno.

### Métodos possíveis
- Leis de Kirchhoff.
- Leis de Newton.
- Conservação de massa.
- Conservação de energia.
- Conservação de quantidade de movimento.
- Relações constitutivas.

### Entregável
Modelo físico conceitual.

---

## 4. Formulação matemática

### Objetivo
Transformar o modelo físico em equações matemáticas.

### Atividades
- Selecionar variáveis dependentes e independentes.
- Formular equações governantes.
- Incorporar condições iniciais e de contorno.
- Verificar consistência dimensional.
- Identificar parâmetros do modelo.

### Tipos de modelos
- Equações algébricas.
- Equações diferenciais ordinárias (EDOs).
- Equações diferenciais parciais (EDPs).
- Equações integrais.
- Sistemas de equações diferenciais.

### Entregável
Modelo matemático contínuo.

---

## 5. Análise matemática do modelo

### Objetivo
Identificar as propriedades que influenciam a escolha
do método numérico.

### Critérios

| Propriedade | Questão |
|---|---|
| Linearidade | O modelo é linear ou não linear? |
| Ordem | Qual é a ordem das derivadas? |
| Dimensionalidade | O problema depende de tempo e/ou espaço? |
| Escalas | Existem escalas de tempo ou comprimento distintas? |
| Rigidez | O sistema apresenta dinâmica rígida? |
| Conservação | Existem quantidades que devem ser preservadas? |
| Regularidade | A solução é suave ou apresenta descontinuidades? |
| Condições | Quais condições iniciais e de contorno são necessárias? |

### Entregável
Caracterização matemática do problema.

---

## 6. Seleção do método numérico

### Objetivo
Selecionar um método compatível com as propriedades do modelo.

### Exemplos

| Classe de problema | Métodos possíveis |
|---|---|
| EDO não rígida | Euler explícito, Runge–Kutta |
| EDO rígida | Euler implícito, BDF |
| EDP de difusão | Diferenças finitas, volumes finitos, elementos finitos |
| Leis de conservação | Volumes finitos, métodos conservativos |
| Problemas estruturais | Elementos finitos |

### Critérios de escolha
- Consistência.
- Estabilidade.
- Convergência.
- Ordem de precisão.
- Custo computacional.
- Preservação das propriedades físicas relevantes.

### Entregável
Método numérico selecionado e justificado.

---

## 7. Discretização

### Objetivo
Transformar o modelo contínuo em um conjunto finito
de equações algébricas.

### Etapas
1. Discretizar o domínio.
2. Definir a malha e seus índices.
3. Aplicar as equações nos pontos ou volumes discretos.
4. Aproximar derivadas e integrais.
5. Incorporar condições iniciais e de contorno.
6. Construir o esquema discreto.

### Exemplo: Forward Euler

Modelo contínuo:

$$
\frac{du}{dt}=f(t,u)
$$

Aproximação da derivada:

$$
\frac{du}{dt}(t_n)
\approx
\frac{u^{n+1}-u^n}{\Delta t}
$$

Equação discreta:

$$
u^{n+1}=u^n+\Delta t f(t_n,u^n)
$$

### Entregáveis
- Malha computacional.
- Esquema discreto.
- Fórmulas de atualização.
- Restrições numéricas conhecidas.

---

## 8. Implementação computacional

### Objetivo
Transformar o esquema numérico em um algoritmo executável.

### Estrutura recomendada

```text
projeto/
├── README.md
├── parameters/
│   └── config.json
├── src/
│   ├── model.py
│   ├── methods.py
│   ├── simulation.py
│   └── analysis.py
├── tests/
│   ├── test_model.py
│   └── test_convergence.py
├── results/
└── notebooks/
```

### Responsabilidades
- `model.py`: equações matemáticas do sistema.
- `methods.py`: algoritmos numéricos.
- `simulation.py`: execução da simulação.
- `analysis.py`: extração das grandezas de interesse.
- `tests/`: testes automatizados.
- `results/`: resultados e gráficos.

A separação entre modelo e método permite utilizar
diferentes discretizações para a mesma equação.

---

## 9. Simulação

### Objetivo
Calcular numericamente a evolução das variáveis do modelo.

### Etapas
1. Carregar parâmetros.
2. Construir a malha.
3. Inicializar as variáveis.
4. Executar o algoritmo numérico.
5. Registrar as soluções.
6. Extrair grandezas físicas.
7. Visualizar os resultados.

### Entregáveis
- Solução numérica.
- Séries temporais ou campos espaciais.
- Gráficos.
- Métricas de interesse.

---

## 10. Verificação numérica

### Pergunta principal
**Estamos resolvendo corretamente as equações discretizadas?**

### Atividades
- Testar a implementação.
- Comparar com soluções analíticas quando disponíveis.
- Realizar refinamento de malha.
- Estimar erros numéricos.
- Verificar estabilidade.
- Estudar convergência.

### Exemplo: convergência

Para um método de ordem p:

$$
E(h)=O(h^p)
$$

Onde h representa o parâmetro de refinamento da malha.

### Entregáveis
- Estimativas de erro.
- Gráficos de convergência.
- Evidências de correção numérica.

---

## 11. Validação física

### Pergunta principal
**O modelo representa adequadamente o fenômeno físico?**

### Atividades
- Comparar com resultados experimentais.
- Comparar grandezas observáveis.
- Avaliar hipóteses do modelo.
- Estudar sensibilidade aos parâmetros.
- Quantificar discrepâncias e incertezas.

### Atenção
Verificação numérica e validação física não são equivalentes.

Uma implementação numericamente correta pode resolver
um modelo fisicamente inadequado.

### Entregável
Avaliação da adequação do modelo ao problema real.

---

## 12. Análise e documentação

### Objetivo
Interpretar resultados e registrar as decisões do projeto.

### Documentar
- Hipóteses físicas.
- Parâmetros e unidades.
- Equações governantes.
- Método numérico escolhido.
- Critérios de estabilidade.
- Refinamentos realizados.
- Resultados e erros.
- Limitações do modelo.
- Conclusões.

## 13. Critério de alinhamento entre modelo e discretização

Antes de iniciar a implementação, responder:

- [ ] As equações governantes foram definidas?
- [ ] As unidades físicas são consistentes?
- [ ] As condições iniciais e de contorno estão completas?
- [ ] As escalas características foram identificadas?
- [ ] O método numérico é compatível com a classe do problema?
- [ ] As condições de estabilidade foram analisadas?
- [ ] Existe uma estratégia de avaliação do erro?
- [ ] Existe uma referência para verificação?
- [ ] Existe uma estratégia de validação física?
- [ ] Os resultados são reproduzíveis?

---

## 14. Princípio central

> O modelo matemático determina quais propriedades físicas
> e matemáticas devem ser respeitadas.
>
> O método numérico define como essas equações serão aproximadas.
>
> A implementação executa o método.
>
> A verificação avalia a correção numérica.
>
> A validação avalia a adequação do modelo ao fenômeno real.
